```python
from datetime import date
from pyspark.sql.types import (StructType, StructField, IntegerType,
                               StringType, DateType, DoubleType)

CATALOGO = "workspace"
ESQUEMA = "ingenieria_datos_juanjo"
TABLA_BRONZE = f"{CATALOGO}.{ESQUEMA}.ventas_bronze"

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOGO}.{ESQUEMA}")

esquema = StructType([
    StructField("id_venta", IntegerType(), True),
    StructField("fecha", DateType(), True),
    StructField("producto", StringType(), True),
    StructField("categoria", StringType(), True),
    StructField("cantidad", IntegerType(), True),
    StructField("precio_unitario", DoubleType(), True),
])

ventas = [
    (1,  date(2026,10,1), "Laptop HP",           "Electrónica", 2,  2800000.0),
    (2,  date(2026,10,1), "Mouse inalámbrico",  "Accesorios",  5,  45000.0),
    (3,  date(2026,10,2), "Teclado mecánico",   "Accesorios",  3,  180000.0),
    (4,  date(2026,10,2), "Monitor 24 pulgadas", "Electrónica", 1,  650000.0),
    (5,  date(2026,10,3), "Camiseta deportiva", "Ropa",        4,  60000.0),
    (5,  date(2026,10,3), "Camiseta deportiva", "Ropa",        4,  60000.0),   # duplicado exacto
    (6,  date(2026,10,3), "Jean clásico",       "Ropa",        2,  120000.0),
    (7,  date(2026,10,4), "Audífonos Bluetooth","Electrónica", 6,  150000.0),
    (8,  date(2026,10,4), "Chaqueta",           "Ropa",        1,  220000.0),
    (9,  date(2026,10,5), None,                 "Accesorios",  2,  35000.0),    # sin producto
    (10, date(2026,10,5), "Cargador USB-C",     "Accesorios",  7,  40000.0),
    (11, date(2026,10,6), "Tablet",             "Electrónica", 1,  900000.0),
    (12, date(2026,10,6), "Gorra",              "Ropa",       -3,  30000.0),    # cantidad negativa
    (13, date(2026,10,7), "Memoria USB 64GB",   "Accesorios",  10, 25000.0),
    (14, date(2026,10,7), "Smartwatch",         "Electrónica", 2,  480000.0),
    (15, date(2026,10,8), "Zapatillas",         "Ropa",        3,  250000.0),
    (16, date(2026,10,8), "Webcam HD",          "Electrónica", 2,  140000.0),
]

df = spark.createDataFrame(ventas, schema=esquema)

# overwrite => ejecutar el notebook varias veces NO acumula filas
(df.write.format("delta")
   .mode("overwrite")
   .option("overwriteSchema", "true")
   .saveAsTable(TABLA_BRONZE))

print(f"Tabla creada: {TABLA_BRONZE} con {spark.table(TABLA_BRONZE).count()} filas")
```
